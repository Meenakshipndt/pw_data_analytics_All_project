## Theortical Question

#### 1. What are the key differences between SQL and NoSQL databases?

```python
SQL is relational, schema-based; NoSQL (like MongoDB) is non-relational, flexible, and scalable.
```

#### 2. What makes MongoDB a good choice for modern applications?

```python
Document-based, flexible schema, high performance, easy scaling, ideal for agile development
```

#### 3. Explain the concept of collections in MongoDB.

```python
Equivalent to tables in SQL; store multiple documents (records) in MongoDB.
```

#### 4. How does MongoDB ensure high availability using replication?

```python
Uses replica sets (1 primary + multiple secondaries) to ensure data redundancy and failover.
```

#### 5. What are the main benefits of MongoDB Atlas?

```python
Fully managed cloud DB, auto-scaling, backups, monitoring, global clusters.
```

#### 6. What is the role of indexes in MongoDB, and how do they improve performance?

```python
Improve query speed; help locate data without scanning entire collection.
```

```python
Stages: $match, $group, $sort, $project, $limit, $lookup, etc.—process data step-by-step.
```

#### 8. What is sharding in MongoDB? How does it differ from replication?

```python
Sharding splits data across servers (scaling); replication copies data (redundancy).
```

#### 9. What is PyMongo, and why is it used?

```python
Python driver for MongoDB—used to interact with MongoDB via Python.
```

#### 10. What are the ACID properties in the context of MongoDB transactions?

```python
Atomicity, Consistency, Isolation, Durability supported via multi-document transactions (from v4.0).
```

#### 11. What is the purpose of MongoDB's explain() function?

```python
Shows how a query is executed (indexes used, scan type, performance insights).
```

#### 12. How does MongoDB handle schema validation?

```python
Defines rules (e.g. required fields, data types) to ensure data quality.
```

#### 13. What is the difference between a primary and a secondary node in a replica set?

```python
Primary handles writes/reads; secondaries replicate data and serve read-only queries.
```

#### 14. What security mechanisms does MongoDB provide for data protection?

```python
Authentication, authorization (roles), encryption (at rest & transit), IP whitelisting.
```

#### 15. Explain the concept of embedded documents and when they should be used.

```python
Store related data inside a document—best for one-to-few relationships and fast reads.
```

#### 16. What is the purpose of MongoDB's $lookup stage in aggregation?

```python
Performs joins between collections (like SQL JOINs).
```

#### 17. What are some common use cases for MongoDB?

```python
Content management, real-time analytics, IoT apps, mobile apps, catalogs, personalization.
```

#### 18. What are the advantages of using MongoDB for horizontal scaling?

```python
Sharding spreads data across nodes—improves performance and storage capacity.
```

#### 19. How do MongoDB transactions differ from SQL transactions?

```python
MongoDB supports multi-document ACID transactions (post-v4.0), but with more flexibility and less locking.
```

#### 20. What are the main differences between capped collections and regular collections?

```python
Capped: fixed size, auto-overwrites old data. Regular: dynamic size, manual deletes.
```

#### 21. What is the purpose of the $match stage in MongoDB's aggregation pipeline?

#### 22. How can you secure access to a MongoDB database?

## Practical Question

#### 1. Write a Python script to load the Superstore dataset from a CSV file into MongoDB.

```python
import pandas as pd
from pymongo import MongoClient

# Load CSV
df = pd.read_csv("Sample - Superstore.csv")  # Adjust path if needed

# Connect to MongoDB
client = MongoClient("mongodb://localhost:27017/")
db = client["superstore"]
collection = db["orders"]

# Insert records
records = df.to_dict(orient="records")
collection.insert_many(records)
```

#### 2. Retrieve and print all documents from the Orders collection.

```python
for doc in collection.find():
    print(doc)
```

#### 3. Count and display the total number of documents in the Orders collection.

```python
print("Total orders:", collection.count_documents({}))
```

#### 4. Write a query to fetch all orders from the "West" region.

```python
for doc in collection.find({"Region": "West"}):
    print(doc)
```

#### 5. Write a query to find orders where Sales is greater than 500.

```python
for doc in collection.find({"Sales": {"$gt": 500}}):
    print(doc)
```

#### 6. Fetch the top 3 orders with the highest Profit.

```python
for doc in collection.find().sort("Profit", -1).limit(3):
    print(doc)
```

#### 7. Update all orders with Ship Mode as "First Class" to "Premium Class."

```python
collection.update_many(
    {"Ship Mode": "First Class"},
    {"$set": {"Ship Mode": "Premium Class"}}
)
```

#### 8. Delete all orders where Sales is less than 50.

```python
collection.delete_many({"Sales": {"$lt": 50}})
```

#### 9. Use aggregation to group orders by Region and calculate total sales per region.

```python
pipeline = [
    {"$group": {"_id": "$Region", "TotalSales": {"$sum": "$Sales"}}}
]

for result in collection.aggregate(pipeline):
    print(result)
```

#### 10. Fetch all distinct values for Ship Mode from the collection.

```python
print(collection.distinct("Ship Mode"))
```

#### 11. Count the number of orders for each category.

```python
pipeline = [
    {"$group": {"_id": "$Category", "OrderCount": {"$sum": 1}}}
]

for result in collection.aggregate(pipeline):
    print(result)
```
